```python
#Funktion zum Abrufen der Daten.
import os
import urllib
import pandas as pd

COUNTRIES_PATH = os.path.join("datasets", "countries")
COUNTRIES_URL = "datasets/countries"

def load_countries_data(countries_path=COUNTRIES_PATH):
    csv_path = os.path.join(countries_path, "countries-aggregated.csv")
    return pd.read_csv(csv_path)
```

```python
#Übersicht über die Variablen, die im Dataframe vorhanden sind.
countries = load_countries_data()
countries.head()
```
